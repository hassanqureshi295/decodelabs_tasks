```python
# ==============================================
# Project 1: Rule-Based AI Chatbot (DecodeLabs)
# ==============================================

import string

responses = {
    "hello": "Hi there! How can I help you today?",
    "hi": "Hello! Nice to see you.",
    "hey": "Hey! What's on your mind?",
    "how are you": "I'm just code, but I'm running perfectly. How about you?",
    "what is your name": "I'm DecodeBot, a rule-based chatbot.",
    "who made you": "I was built by an AI engineering intern at DecodeLabs.",
    "what can you do": "I can chat with you using predefined rules. Try greetings, or ask my name.",
    "help": "Try saying: hello, how are you, what is your name, what can you do, or thanks.",
    "thanks": "You're welcome!",
    "thank you": "Happy to help!",
}

exit_commands = {"exit", "quit", "bye", "goodbye"}

FALLBACK = "I do not understand. Type 'help' to see what I can answer."


def sanitize(text):
    """Phase 1: Input and Sanitization - fix case, whitespace, and punctuation."""
    text = text.lower().strip()
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = " ".join(text.split())
    return text


def chatbot():
    print("DecodeBot: Hello! Type 'exit' to end the chat.\n")

    while True:
        raw_input = input("You: ")
        clean_input = sanitize(raw_input)

        if clean_input in exit_commands:
            print("DecodeBot: Goodbye! Have a great day.")
            break

        if clean_input == "":
            print("DecodeBot: Please type something.")
            continue

        reply = responses.get(clean_input, FALLBACK)
        print("DecodeBot:", reply)


chatbot()
```
